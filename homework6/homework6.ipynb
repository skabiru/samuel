```python
arr = np.array([[2, 3, 5],[14, 6, 8],[11, 13, 17],[7, 10, 13]])
def rows_prime(arr):
    def prime(num):
        if num < 2:
            return False
        for i in range(2, num):
            if num % i == 0:
                return False
        return True

    result = []
    for row in arr:
        for val in row:
            if prime(val):
                result.append(row)
    return np.array(result)
```

```python
def checkerboard():
    board = np.zeros((8,8))
    board[::2, 1::2] = 1 
    board[1::2, ::2] = 1
    return 1 - board
def reverse_checkerboard():
    board = np.zeros((8,8))
    board[::2, 1::2] = 1 
    board[1::2, ::2] = 1
    return board
```

```python
def expansion(string, n):
    result = []
    for word in string:
        spaces = (" " * n).join(word)
        result.append(spaces)
    return np.array(result)
```

```python
def second_brightest(arr):
    result = []
    for row in arr:
        sorted_row = np.sort(row)
        result.append(sorted_row[-2])
    return np.array(result)
```
