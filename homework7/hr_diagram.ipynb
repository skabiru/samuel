```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("../../course_assignments/homework7/star_data.txt",skiprows=1)
```

```python
print(data.shape) # there are 2642 rows
magnitude = data[:, 0]
temperature = data[:, 1]
min_mag = np.min(magnitude)
max_mag = np.max(magnitude)
min_mag_star = np.argmin(magnitude)
max_mag_star = np.argmax(magnitude)
min_temp = np.min(temperature)
max_temp = np.max(temperature)
min_temp_star = np.argmin(temperature)
max_temp_star = np.argmax(temperature)
dimmest_star = max_mag_star
brighest_star = min_mag_star
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
fig = plt.figure(figsize=(8,6))
plt.scatter(log_temperature, log_magnitude, c = log_temperature, cmap = "plasma")
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
plt.scatter(log_temperature[brighest_star], log_magnitude[brighest_star], color = 'black', marker = "*", s = 300, label = "brighest star")
plt.scatter(log_temperature[dimmest_star], log_magnitude[dimmest_star], color = 'purple', marker = "*", s = 300, label = "dimmest star")
plt.scatter(log_temperature[max_temp_star], log_magnitude[max_temp_star], color = 'green', marker = "*", s = 300, label = "hottest star,")
plt.scatter(log_temperature[min_temp_star], log_magnitude[min_temp_star], color = 'orange', marker = "*", s = 300, label = "coldest star")
plt.legend()
plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")
plt.show()
```
