```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Matplotlib
def make_sine_wave(x, A, w):
    return A * np.sin(w*x)
x = np.linspace(0,2 * np.pi,1000)
amplitude = np.array([0.5, 1, 1.5, 2, 2.5])
frequency = np.array([1, 2, 3, 4, 5])
fig, axes = plt.subplots(1, 3, figsize = (25, 5))
style = np.array(['-', '--', ':', '-.'])
for A, w , s in zip(amplitude, frequency, style):
    y1 = make_sine_wave(x, A, 1)
    axes[0].plot(x, y1, linestyle = s, label = f"A = {A}, w = {w}")
    y2 = make_sine_wave(x, 1, w)
    axes[1].plot(x, y2, linestyle = s, label = f"A = {A}, w = {w}")
    y3 = make_sine_wave(x, A, w)
    axes[2].plot(x, y3, linestyle = s, label = f"A = {A}, w = {w}")
    
axes[0].set_title("Varying Amplitude", fontsize = 24)
axes[0].set_xlabel("Time", fontsize = 20)
axes[0].set_ylabel("Amplitude", fontsize = 20)
axes[0].legend()
axes[1].set_title("Varying Frequency", fontsize = 24)
axes[1].set_xlabel("Time", fontsize = 20)
axes[1].set_ylabel("Amplitude", fontsize = 20)
axes[1].legend()
axes[2].set_title("Varying Amplitude and Frequency", fontsize = 24)
axes[2].set_xlabel("Time", fontsize = 20)
axes[2].set_ylabel("Amplitude", fontsize = 20)
axes[2].legend()
plt.show()
```
